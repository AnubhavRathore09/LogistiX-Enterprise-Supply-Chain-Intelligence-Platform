# Business Feature Engineering

The following section creates business-driven analytical features that improve supply chain performance analysis, executive reporting, KPI calculation, and dashboard insights.

```python
df.columns.tolist()
```

```python
feature_df["order_date_dateorders"] = pd.to_datetime(
    feature_df["order_date_dateorders"]
)

feature_df["shipping_date_dateorders"] = pd.to_datetime(
    feature_df["shipping_date_dateorders"]
)
```

```python
feature_df["shipping_lead_time"] = (
    feature_df["shipping_date_dateorders"]
    - feature_df["order_date_dateorders"]
).dt.days
```

```python
feature_df["delivery_delay"] = (
    feature_df["days_for_shipping_real"]
    - feature_df["days_for_shipment_scheduled"]
)
```

```python
feature_df["on_time_delivery"] = np.where(
    feature_df["delivery_delay"] <= 0,
    "On Time",
    "Delayed"
)
```

```python
feature_df["profit_margin"] = (
    feature_df["order_profit_per_order"]
    / feature_df["sales"]
) * 100
```

```python
feature_df["order_value_category"] = pd.qcut(
    feature_df["sales"],
    q=4,
    labels=[
        "Low",
        "Medium",
        "High",
        "Premium"
    ]
)
```

```python
feature_df["customer_type"] = np.where(
    feature_df["customer_segment"] == "Consumer",
    "B2C",
    "B2B"
)
```

```python
feature_df["high_value_order"] = np.where(
    feature_df["sales"] >= feature_df["sales"].median(),
    1,
    0
)
```

```python
feature_df["weekend_order"] = np.where(
    feature_df["order_date_dateorders"].dt.dayofweek >= 5,
    1,
    0
)
```

```python
feature_df["order_month"] = (
    feature_df["order_date_dateorders"]
    .dt.month_name()
)
```

```python
feature_df["order_year"] = (
    feature_df["order_date_dateorders"]
    .dt.year
)
```

```python
feature_df["shipping_speed"] = pd.cut(
    feature_df["days_for_shipping_real"],
    bins=[0,2,5,20],
    labels=[
        "Fast",
        "Standard",
        "Slow"
    ]
)
```

```python
feature_df["profit_category"] = np.where(
    feature_df["order_profit_per_order"] >= 0,
    "Profit",
    "Loss"
)
```

```python
feature_df["discount_category"] = pd.cut(
    feature_df["order_item_discount_rate"],
    bins=[-0.01,0,0.10,0.25,1],
    labels=[
        "No Discount",
        "Low",
        "Medium",
        "High"
    ]
)
```

```python
feature_df["quantity_category"] = pd.cut(
    feature_df["order_item_quantity"],
    bins=[0,1,3,5,100],
    labels=[
        "Single",
        "Small",
        "Medium",
        "Bulk"
    ]
)
```

```python
new_features = [
    "shipping_lead_time",
    "delivery_delay",
    "on_time_delivery",
    "profit_margin",
    "order_value_category",
    "customer_type",
    "high_value_order",
    "weekend_order",
    "order_month",
    "order_year",
    "shipping_speed",
    "profit_category",
    "discount_category",
    "quantity_category"
]

feature_df[new_features].head()
```
