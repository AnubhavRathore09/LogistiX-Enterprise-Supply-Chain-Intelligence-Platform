```python
import pandas as pd

pd.set_option("display.max_columns", None)
```

```python
df = pd.read_csv(
    "../../Dataset/Raw/DataCoSupplyChainDataset.csv",
    encoding="latin1"
)

df.shape
```

```python
summary = pd.DataFrame({

    "KPI":[
        "Total Sales",
        "Total Orders",
        "Total Customers",
        "Total Profit",
        "Markets",
        "Departments"
    ],

    "Value":[
        f"£{df['Sales'].sum():,.2f}",
        df["Order Id"].nunique(),
        df["Customer Id"].nunique(),
        f"£{df['Order Profit Per Order'].sum():,.2f}",
        df["Market"].nunique(),
        df["Department Name"].nunique()
    ]

})

summary
```
