```python
import warnings
warnings.filterwarnings("ignore")

import pandas as pd
import numpy as np

pd.set_option("display.max_columns", None)
pd.set_option("display.width", 1000)

print("LogistiX - Data Cleaning Started")
```

```python
file_path = "../data/raw/DataCoSupplyChainDataset/DataCoSupplyChainDataset.csv"

df = pd.read_csv(file_path, encoding="latin1")

print(df.shape)
```

```python
clean_df = df.copy()

print(clean_df.shape)
```

```python
clean_df.head()
```

```python
clean_df.columns.tolist()
```

```python
clean_df.columns = (
    clean_df.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
    .str.replace("(", "", regex=False)
    .str.replace(")", "", regex=False)
    .str.replace("-", "_")
)
```

```python
clean_df.columns.tolist()
```

```python
clean_df.info()
```

```python
clean_df.isnull().sum().sort_values(ascending=False)
```

```python
(clean_df.isnull().sum()/len(clean_df)*100).round(2).sort_values(ascending=False)
```

```python
clean_df.duplicated().sum()
```

```python
clean_df = clean_df.drop_duplicates()

print(clean_df.shape)
```

```python
clean_df.select_dtypes(include="object").columns
```

```python
object_columns = clean_df.select_dtypes(include="object").columns

clean_df[object_columns] = clean_df[object_columns].apply(
    lambda x: x.str.strip() if x.dtype == "object" else x
)
```

```python
clean_df.head()
```

```python
clean_df.select_dtypes(include=["int64","float64"]).describe().T
```

```python
date_columns = [col for col in clean_df.columns if "date" in col]

date_columns
```

```python
for col in date_columns:
    clean_df[col] = pd.to_datetime(clean_df[col], errors="coerce")
```

```python
clean_df[date_columns].dtypes
```

```python
clean_df.isnull().sum().sort_values(ascending=False)
```

```python
clean_df.describe(include="all").T
```

```python
clean_df.shape
```

```python
clean_df.info()
```

```python
clean_df.to_csv(
    "../data/processed/cleaned_supply_chain_dataset.csv",
    index=False
)
```

```python
print("Cleaned dataset exported successfully.")
```
