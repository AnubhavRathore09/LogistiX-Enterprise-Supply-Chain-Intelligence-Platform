# Column Understanding

The objective of this section is to understand the business meaning of every column in the dataset before performing any transformations. This helps ensure accurate data cleaning, feature engineering, SQL analysis, and dashboard development.

```python
pd.DataFrame({
    "Column Name": df.columns,
    "Data Type": df.dtypes.values,
    "Missing Values": df.isnull().sum().values,
    "Unique Values": df.nunique().values
})
```

```python
column_summary = pd.DataFrame({
    "Column Name": df.columns,
    "Data Type": df.dtypes.values,
    "Missing Values": df.isnull().sum().values,
    "Missing Percentage": (df.isnull().sum()/len(df)*100).round(2).values,
    "Unique Values": df.nunique().values
})

column_summary
```

```python
column_summary.sort_values("Missing Percentage", ascending=False)
```

```python
column_summary.sort_values("Unique Values")
```

# Dataset Quality Summary

This summary provides an overview of the dataset quality before any preprocessing activities.

```python
quality_report = pd.DataFrame({
    "Metric": [
        "Rows",
        "Columns",
        "Duplicate Rows",
        "Total Missing Values",
        "Numerical Columns",
        "Categorical Columns"
    ],
    "Value": [
        len(df),
        len(df.columns),
        df.duplicated().sum(),
        df.isnull().sum().sum(),
        len(df.select_dtypes(exclude="object").columns),
        len(df.select_dtypes(include="object").columns)
    ]
})

quality_report
```
